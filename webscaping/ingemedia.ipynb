```python
import pandas as pd 

xls = pd.ExcelFile('etudiants-ingemedia.xlsx')

df_2008 = pd.read_excel(xls, sheet_name='2007-2008')
df_2008['promotion'] = '2008'
df_2009 = pd.read_excel(xls, sheet_name='2008-2009')
df_2009['promotion'] = '2009'
df_2010 = pd.read_excel(xls, sheet_name='2009-2010')
df_2010['promotion'] = '2010'
df_2011 = pd.read_excel(xls, sheet_name='2010-2011')
df_2011['promotion'] = '2011'
df_2012 = pd.read_excel(xls, sheet_name='2011-2012')
df_2012['promotion'] = '2012'
df_2013 = pd.read_excel(xls, sheet_name='2012-2013')
df_2013['promotion'] = '2013'
df_2014 = pd.read_excel(xls, sheet_name='2013-2014')
df_2014['promotion'] = '2014'
df_2015 = pd.read_excel(xls, sheet_name='2014-2015')
df_2015['promotion'] = '2015'
df_2016 = pd.read_excel(xls, sheet_name='2015-2016')
df_2016['promotion'] = '2016'
df_2017 = pd.read_excel(xls, sheet_name='2016-2017')
df_2017['promotion'] = '2017'
df_2018 = pd.read_excel(xls, sheet_name='2017-2018')
df_2018['promotion'] = '2018'
df_2019 = pd.read_excel(xls, sheet_name='2018-2019')
df_2019['promotion'] = '2019'
df_2020 = pd.read_excel(xls, sheet_name='2019-2020')
df_2020['promotion'] = '2020'
df_2021 = pd.read_excel(xls, sheet_name='2020-2021')
df_2021['promotion'] = '2021'
df_2022 = pd.read_excel(xls, sheet_name='2021-2022')
df_2022['promotion'] = '2022'
df_2023 = pd.read_excel(xls, sheet_name='2022-2023')
df_2023['promotion'] = '2023'
df_2024 = pd.read_excel(xls, sheet_name='2023-2024')
df_2024['promotion'] = '2024'
```

```python
df_all = pd.concat([df_2008, df_2009, df_2010, df_2011, df_2012, df_2013, df_2014, df_2015, df_2016, df_2017, df_2018, df_2019, df_2020, df_2021, df_2022, df_2023, df_2024], ignore_index=True)
df_all.head(10)
```

```python
df_all_cleaned = df_all.copy()
df_all_cleaned = df_all_cleaned.drop(columns=['Formation'])
# Prendre la première valeur non-nulle entre les deux colonnes
df_all_cleaned['nom_entier'] = df_all_cleaned['etudiants_nom'].fillna(df_all_cleaned['nom_etudiant'])

# Supprimer tout ce qui est entre parenthèses (ex: "(NTS)")
df_all_cleaned['nom_entier'] = df_all_cleaned['nom_entier'].str.replace(r'\s*\([^)]*\)', '', regex=True).str.strip()


# Si virgule : séparer par ',', sinon par espace
df_all_cleaned['nom'] = df_all_cleaned['nom_entier'].apply(
    lambda x: x.split(',')[0].strip() if ',' in str(x) else str(x).split(' ')[-1]
)

df_all_cleaned['prenom'] = df_all_cleaned['nom_entier'].apply(
    lambda x: x.split(',')[1].strip() if ',' in str(x) else str(x).split(' ')[0]
)

df_all_cleaned = df_all_cleaned.drop(columns=['etudiants_nom', 'nom_etudiant','intitule_etudiant','nom_entier'])

df_all_cleaned.head(10)
```
