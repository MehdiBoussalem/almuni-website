```python
import pandas as pd
csv = "alumni liste.csv"
json ="result alumnis.json"
df = pd.read_csv(csv,sep=",")

# Create list of schools
df['Liste des écoles'] = df.apply(lambda x: ', '.join(filter(pd.notna, [x.get('linkedinSchoolName'), x.get('linkedinPreviousSchoolName')])), axis=1)

# Create list of jobs
df['Liste des jobs'] = df.apply(lambda x: ', '.join(filter(pd.notna, [x.get('linkedinJobTitle'), x.get('linkedinPreviousJobTitle')])), axis=1)

# Select and rename columns
columns_to_keep = {
    'lastName': 'Nom',
    'firstName': 'Prénom',
    'profileUrl': 'URL LinkedIn',
    'linkedinProfileImageUrl': 'Photo URL',
    'Liste des écoles': 'Liste des écoles',
    'Liste des jobs': 'Liste des jobs',
    'linkedinJobTitle': 'Poste actuel'
}

final_df = df[list(columns_to_keep.keys())].rename(columns=columns_to_keep)
final_df.head()
```

```python
df['linkedinDescription'].head() 
X = ['firstName', 'lastName', 'linkedinJobTitle','location','linkedinSchoolName','linkedinSchoolDateRange','linkedinProfileUrl','linkedinProfileImageUrl']
new_df = df[X]
new_df.to_csv('alumni.csv', index=False)
```

```python
print(f' All : {len(new_df)}')
ingemedia_df = new_df[new_df['linkedinSchoolName'].isin(['Université de Toulon', 'Institut Ingemedia'])].copy()
ingemedia_df["promo"] = ingemedia_df["linkedinSchoolDateRange"].str[-4:]
ingemedia_df = ingemedia_df[pd.to_numeric(ingemedia_df["promo"], errors='coerce').fillna(9999) < 2026]
ingemedia_df.drop(columns=['linkedinSchoolDateRange','linkedinSchoolName'], inplace=True)
ingemedia_df.head()
```

```python
ingemedia_df = ingemedia_df.rename(columns={
    "lastName": "nom",
    "firstName": "prenom",
    "location": "lieu",
    "linkedinJobTitle": "poste",
    "linkedinPhoto": "url_photo",
    "promo": "promo",
    "linkedinProfileUrl": "profil_url"
})
ingemedia_df.head()
```

```python
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter
import time

# Initialiser le géocodeur (ajoutez un user_agent unique)
geolocator = Nominatim(user_agent="ingemedia_alumni_app")

# Rate limiter pour éviter de surcharger le service (1 requête par seconde)
geocode = RateLimiter(geolocator.geocode, min_delay_seconds=1)

def get_coordinates(lieu):
    try:
        if pd.isna(lieu) or lieu == "":
            return None, None
        location = geocode(lieu)
        if location:
            return location.latitude, location.longitude
        return None, None
    except Exception as e:
        print(f"Erreur pour '{lieu}': {e}")
        return None, None

# Appliquer la fonction sur la colonne 'lieu'
ingemedia_df[["latitude", "longitude"]] = ingemedia_df["lieu"].apply(
    lambda x: pd.Series(get_coordinates(x))
)

ingemedia_df.head()
```
