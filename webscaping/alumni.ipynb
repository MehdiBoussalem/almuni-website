```python
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter
import time

# Initialiser le géocodeur (ajoutez un user_agent unique)
geolocator = Nominatim(user_agent="ingemedia_alumni_app")

# Rate limiter pour éviter de surcharger le service (1 requête par seconde)
geocode = RateLimiter(geolocator.geocode, min_delay_seconds=1)

def get_coordinates(lieu):
    try:
        if pd.isna(lieu) or lieu == "":
            return None, None
        location = geocode(lieu)
        if location:
            return location.latitude, location.longitude
        return None, None
    except Exception as e:
        print(f"Erreur pour '{lieu}': {e}")
        return None, None

# Appliquer la fonction sur la colonne 'lieu'
ingemedia_df[["latitude", "longitude"]] = ingemedia_df["lieu_clean"].apply(
    lambda x: pd.Series(get_coordinates(x))
)
ingemedia_df.head()
```

```python
# afficher la liste des nulls dans latitude et longitude
null_coords = ingemedia_df[ingemedia_df['latitude'].isnull() | ingemedia_df['longitude'].isnull()]
print("Lieux sans coordonnées géographiques:")
print(null_coords['lieu'].unique())
```

```python
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter

geolocator = Nominatim(user_agent="ingemedia_alumni_app")
reverse_geocode = RateLimiter(geolocator.reverse, min_delay_seconds=1)

def get_city_from_coordinates(lat, lon):
    try:
        if pd.isna(lat) or pd.isna(lon):
            return None
        location = reverse_geocode(f"{lat}, {lon}")
        if location:
            # Extraire la ville de l'adresse
            address = location.raw['address']
            city = address.get('city') or address.get('town') or address.get('village')
            return city
        return None
    except Exception as e:
        print(f"Erreur pour ({lat}, {lon}): {e}")
        return None

# Créer la colonne ville
ingemedia_df['ville'] = ingemedia_df.apply(
    lambda row: get_city_from_coordinates(row['latitude'], row['longitude']), 
    axis=1
)

ingemedia_df[['lieu_clean', 'latitude', 'longitude', 'ville']].head()
```
